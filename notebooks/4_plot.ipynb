# Plotting of Time Series

## Loading Data

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df = pd.read_csv('../data/csv/cars_self_contained.csv', parse_dates=['date'])

df = df[df['locationName'] == "Brugg"]

df = df.sort_values('date')
```

## Seaborn Barplot

This is very much the figure, that https://visualize.admin.ch is also able to draw right out of the box.

```python
ig, ax = plt.subplots(figsize=(12, 6))  # width=12, height=6

ax = sns.barplot(data=df, x='date', y='PERS_WAGEN', hue='change_event_name', dodge=False, width=0.6, ax = ax)

ax.yaxis.grid(True)   # horizontal lines
ax.xaxis.grid(False)  # turn off vertical grid if any

# Put grid behind bars
ax.set_axisbelow(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Seaborn Lineplot

```python
ig, ax = plt.subplots(figsize=(12, 6))  # width=12, height=6

ax = sns.lineplot(data=df, x='date', y='PERS_WAGEN', hue='change_event_name', ax = ax)

ax.yaxis.grid(True)   # horizontal lines
ax.xaxis.grid(False)  # turn off vertical grid if any

# Put grid behind bars
ax.set_axisbelow(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
