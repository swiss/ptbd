# How to Use LINDAS data

This notebook shows how to use the LINDAS data for integrating it into other systems and projects. The following three methods are covered:

- Download the data as CSV file via html link or `curl` command
- Get JSON data via API
- Use API and some scripting to get a pandas DataFrame directly in Python

## Query and Parameters

The following query and and endpoint parameter are used for all three methods:

```python
QUERY = """

PREFIX schema: <http://schema.org/>
PREFIX vl: <https://version.link/>
PREFIX admin: <https://schema.ld.admin.ch/>

SELECT ?name ?validFrom ?changeEventDescription WHERE {

    ?muni a admin:PoliticalMunicipality ;
        schema:name ?name ;
        schema:isPartOf/schema:isPartOf <https://ld.admin.ch/canton/10> ;
        vl:version ?version .
    ?version schema:validFrom ?validFrom ;
        vl:startEvent/admin:changeEventDescription/schema:name ?changeEventDescription .

    FILTER(NOT EXISTS {?muni a vl:Deprecated})
    FILTER (lang(?changeEventDescription) = 'en')

} ORDER BY DESC(?validFrom)

"""
```

```python
ENDPOINT = "https://int.ld.admin.ch/query"
```

## Download Data as CSV

### URL Encoding

```python
from urllib.parse import quote

query_url_encoded = quote(QUERY)

print(query_url_encoded)
```

### Clickable Link

```python
clickable_link = f"{ENDPOINT}?format=csv&query={query_url_encoded}"

print(clickable_link) 
```

### Programmatically Download CSV with CURL

```python
import subprocess

subprocess.run(["curl", "-L", "-o", "../data/csv/muni_FR.csv", clickable_link])

# actual use on the command line would be:
# curl -L -o ../data/csv/muni_FR.csv "https://int.ld.admin.ch/query?format=csv&query=PREFIX%20schema%3A%20...
```

## Get JSON Data via API

```python
import httpx

def sparql_query(query: str) -> dict:
    with httpx.Client(timeout=30.0) as client:
        response = client.get(
            ENDPOINT,
            params={"query": query},
            headers={"Accept": "application/sparql-results+json"},
        )
        response.raise_for_status()
        return response.json()
    
result = sparql_query(QUERY)
```

```python
import json

print(json.dumps(result, indent=2, ensure_ascii=False))
```

## Results as Pandas DataFrame

```python
import pandas as pd

bindings = result["results"]["bindings"]
# Convert the JSON results to a pandas DataFrame but use only the values of the bindings, not the full structure
df = pd.json_normalize(bindings)

display(df)
```
